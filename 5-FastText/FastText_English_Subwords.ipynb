```python
def generate_subwords(word, n_min=3, n_max=6):
    """
    Generate FastText-style subwords for a word.

    Parameters:
        word (str): Input word
        n_min (int): Minimum n-gram length
        n_max (int): Maximum n-gram length

    Returns:
        dict: Contains boundary-marked word, extracted n-grams, and final token set
    """
    marked_word = f"<{word}>"
    ngrams = []

    for n in range(n_min, n_max + 1):
        for i in range(len(marked_word) - n + 1):
            ngrams.append(marked_word[i:i+n])

    final_tokens = sorted(set(ngrams + [word]))

    return {
        "word": word,
        "marked_word": marked_word,
        "ngrams": ngrams,
        "final_tokens": final_tokens
    }


# Example English words
english_words = ["fast", "walk", "walking", "walked", "breathtakingly"]

for w in english_words:
    result = generate_subwords(w, n_min=3, n_max=6)

    print("=" * 60)
    print("Word:", result["word"])
    print("Boundary-marked word:", result["marked_word"])
    print("Extracted n-grams:")
    print(result["ngrams"])
    print("Final token set (including full word):")
    print(result["final_tokens"])
    print()
```

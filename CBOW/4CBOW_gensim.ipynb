```python
from gensim.models import Word2Vec

corpus = "The quick brown fox jumps over the lazy dog"

# Tokenize and normalize
tokens = corpus.lower().split()

# Create vocabulary information
vocabulary = list(dict.fromkeys(tokens))
V = len(vocabulary)  # Vocabulary size
N = 3                # Desired embedding dimension

print("CBOW Word2Vec Example-Single Sentence")
print("STEP 1: Vocabulary")
print("Vocabulary:", vocabulary)
print("Vocabulary size V =", V)
print("Embedding dimension N =", N)

# Word2Vec expects a list of sentences,
# where each sentence is a list of words
sentences = [tokens]
print("sentences:", sentences)
# Train the Word2Vec model using CBOW (sg=0)
model = Word2Vec(
    sentences,
    vector_size=N,
    window=2,
    min_count=1,
    sg=0
)

# Get the vector representation of a word
vector = model.wv["fox"]

print("\nWords learned by model:")
print(model.wv.index_to_key)

print("\nSTEP 2:Words learned:")
print(model.wv.index_to_key)

# Get vector for a word
vector = model.wv["fox"]

print("\nSTEP 3:Vector representation of 'fox':")
print(vector)

print("\nSTEP 4:Similar words to 'fox':")
similar_words = model.wv.most_similar("fox", topn=3)
print(similar_words)
print("\nSTEP 5:Similarity between 'fox' and 'jumps':")
similarity = model.wv.similarity("fox", "jumps")
print(similarity)
```

```python
from gensim.models import Word2Vec

corpus = [
    ['the', 'quick', 'brown', 'fox', 'jumps', 'over', 'the', 'lazy', 'dog'],
    ['the', 'dog', 'chased', 'the', 'cat'],
    ['the', 'fox', 'ran', 'into', 'the', 'forest'],
    ['the', 'cat', 'climbed', 'a', 'tree']
]

# Create vocabulary information
vocabulary = sorted(set(word for sentence in corpus for word in sentence))
V = len(vocabulary)
N = 3

print("STEP 1: Vocabulary")
print("Vocabulary:", vocabulary)
print("Vocabulary size V =", V)
print("Embedding dimension N =", N)

# Train CBOW model (sg=0)
model = Word2Vec(
    corpus,
    vector_size=N,
    window=2,
    min_count=1,
    sg=0
)

# Display learned words
print("\nSTEP 2:Words learned:")
print(model.wv.index_to_key)

# Get vector for a word
vector = model.wv["fox"]

print("\nSTEP 3:Vector representation of 'fox':")
print(vector)

print("\nSTEP 4:Similar words to 'tree':")
similar_words = model.wv.most_similar("tree")
print(similar_words)
print("\nSTEP 5:Similarity between 'cat' and 'dog':")
similarity = model.wv.similarity("cat", "dog")
print(similarity)
```
